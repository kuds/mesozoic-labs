## Curriculum Stage Configurations

SAC uses the same curriculum as PPO but with SAC-specific hyperparameters.

```python
STAGE_CONFIGS = {
    1: {
        "name": "balance",
        "description": "Learn to stand without falling",
        "env_kwargs": {
            "forward_vel_weight": 0.0,
            "alive_bonus": 1.0,
            "energy_penalty_weight": 0.0005,
            "tail_stability_weight": 0.1,
            "bite_bonus": 0.0,
            "bite_approach_weight": 0.0,
            "prey_distance_range": (10.0, 15.0),
            "max_episode_steps": 500,
        },
        "sac_kwargs": {
            "learning_rate": 3e-4,
            "batch_size": 256,
            "gamma": 0.99,
            "tau": 0.005,
            "ent_coef": "auto",
            "buffer_size": 1_000_000,
            "learning_starts": 10_000,
        },
        "timesteps": 500_000,
    },
    2: {
        "name": "locomotion",
        "description": "Walk and run forward",
        "env_kwargs": {
            "forward_vel_weight": 1.0,
            "alive_bonus": 0.5,
            "energy_penalty_weight": 0.001,
            "tail_stability_weight": 0.05,
            "bite_bonus": 0.0,
            "bite_approach_weight": 0.2,
            "prey_distance_range": (8.0, 12.0),
            "max_episode_steps": 1000,
        },
        "sac_kwargs": {
            "learning_rate": 1e-4,
            "batch_size": 256,
            "gamma": 0.99,
            "tau": 0.005,
            "ent_coef": "auto",
            "buffer_size": 1_000_000,
            "learning_starts": 10_000,
        },
        "timesteps": 1_000_000,
    },
    3: {
        "name": "bite",
        "description": "Sprint and bite prey with jaw",
        "env_kwargs": {
            "forward_vel_weight": 1.0,
            "alive_bonus": 0.1,
            "energy_penalty_weight": 0.001,
            "tail_stability_weight": 0.02,
            "bite_bonus": 500.0,
            "bite_approach_weight": 0.5,
            "prey_distance_range": (3.0, 8.0),
            "prey_lateral_range": (-1.5, 1.5),
            "max_episode_steps": 1000,
        },
        "sac_kwargs": {
            "learning_rate": 5e-5,
            "batch_size": 512,
            "gamma": 0.995,
            "tau": 0.005,
            "ent_coef": "auto",
            "buffer_size": 2_000_000,
            "learning_starts": 10_000,
        },
        "timesteps": 2_000_000,
    },
}

for stage, config in STAGE_CONFIGS.items():
    print(f"Stage {stage}: {config['name']} - {config['description']}")
    print(f"  Timesteps: {config['timesteps']:,}")
    print()
```

## Training Helpers

```python
N_ENVS = 4
SEED = 42
log_dir = "./logs/trex_sac"


def make_env(stage, rank, seed=0):
    """Create a single TRexEnv instance."""

    def _init():
        env_kwargs = STAGE_CONFIGS[stage]["env_kwargs"].copy()
        env = TRexEnv(**env_kwargs)
        env = Monitor(env)
        env.reset(seed=seed + rank)
        return env

    set_random_seed(seed)
    return _init


def create_vec_env(stage, n_envs=N_ENVS, seed=SEED):
    """Create vectorized environment with observation/reward normalization."""
    env = DummyVecEnv([make_env(stage, i, seed) for i in range(n_envs)])
    env = VecNormalize(env, norm_obs=True, norm_reward=True, clip_obs=10.0, clip_reward=10.0)
    return env


print("Training utilities ready.")
```

## Train

Set `STAGE` to 1, 2, or 3. For quick testing set `QUICK_TEST = True`.

```python
STAGE = 1
QUICK_TEST = True
TIMESTEPS = 500_000 if QUICK_TEST else STAGE_CONFIGS[STAGE]["timesteps"]

print(f"Training Stage {STAGE}: {STAGE_CONFIGS[STAGE]['name']}")
print(f"Timesteps: {TIMESTEPS:,}")

# Create environments
train_env = create_vec_env(STAGE)
eval_env = create_vec_env(STAGE, n_envs=1, seed=SEED + 1000)

# Create SAC model
config = STAGE_CONFIGS[STAGE]
sac_kwargs = config["sac_kwargs"].copy()
sac_kwargs["verbose"] = 1

model = SAC("MlpPolicy", train_env, **sac_kwargs)

# Callbacks
stage_dir = os.path.join(log_dir, f"stage{STAGE}_{datetime.now().strftime('%Y%m%d_%H%M%S')}")
os.makedirs(stage_dir, exist_ok=True)
model_dir = os.path.join(stage_dir, "models")
os.makedirs(model_dir, exist_ok=True)

eval_callback = EvalCallback(
    eval_env,
    best_model_save_path=model_dir,
    log_path=stage_dir,
    eval_freq=max(5000 // N_ENVS, 1),
    n_eval_episodes=3,
    deterministic=True,
)

checkpoint_callback = CheckpointCallback(
    save_freq=max(10000 // N_ENVS, 1),
    save_path=model_dir,
    name_prefix=f"stage{STAGE}",
)

# Train
print(f"\nTraining for {TIMESTEPS:,} timesteps...")
model.learn(
    total_timesteps=TIMESTEPS,
    callback=CallbackList([eval_callback, checkpoint_callback]),
    progress_bar=True,
)

# Save
final_path = os.path.join(model_dir, f"stage{STAGE}_final")
model.save(final_path)
train_env.save(final_path + "_vecnorm.pkl")
print(f"\nModel saved to: {final_path}.zip")

# Evaluate
mean_reward, std_reward = evaluate_policy(model, eval_env, n_eval_episodes=5)
print(f"Mean reward: {mean_reward:.2f} +/- {std_reward:.2f}")

train_env.close()
eval_env.close()
```

## Training Curve

```python
eval_log = os.path.join(stage_dir, "evaluations.npz")

if os.path.exists(eval_log):
    data = np.load(eval_log)
    timesteps = data["timesteps"]
    results = data["results"]
    mean_rewards = np.mean(results, axis=1)
    std_rewards = np.std(results, axis=1)

    plt.figure(figsize=(10, 5))
    plt.plot(timesteps, mean_rewards, "r-", label="Mean Reward")
    plt.fill_between(timesteps, mean_rewards - std_rewards, mean_rewards + std_rewards, alpha=0.3, color="red")
    plt.xlabel("Timesteps")
    plt.ylabel("Reward")
    plt.title(f"T-Rex SAC - Stage {STAGE} Training Progress")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
else:
    print("No evaluation log found.")
```

## Record Final Model Video

Render one episode of the trained policy and save as MP4.

```python
import mediapy


def make_render_env(stage, seed=0):
    """Create a single TRexEnv with rgb_array rendering for video capture."""

    def _init():
        env_kwargs = STAGE_CONFIGS[stage]["env_kwargs"].copy()
        env = TRexEnv(render_mode="rgb_array", **env_kwargs)

        # ZOOM IN: Lower the Field of View (FOV)
        # Even lower FOV for more zoom (was 20)
        env.model.vis.global_.fovy = 15
        for i in range(env.model.ncam):
            env.model.cam_fovy[i] = 15

        # BRIGHTER LIGHTING
        env.model.vis.headlight.diffuse[:] = [0.8, 0.8, 0.8]
        env.model.vis.headlight.ambient[:] = [0.3, 0.3, 0.3]
        env.model.vis.headlight.specular[:] = [0.1, 0.1, 0.1]

        env = Monitor(env)
        env.reset(seed=seed)
        return env

    return _init


# Build env with rendering and load normalization stats from training
video_env = DummyVecEnv([make_render_env(STAGE, seed=SEED + 2000)])
video_env = VecNormalize.load(final_path + "_vecnorm.pkl", video_env)
video_env.training = False
video_env.norm_reward = False

obs = video_env.reset()
frames = []
episode_reward = 0.0

for _ in range(1000):
    action, _ = model.predict(obs, deterministic=True)
    obs, reward, done, info = video_env.step(action)
    frames.append(video_env.venv.envs[0].render())
    episode_reward += reward[0]
    if done[0]:
        break

video_env.close()

# Save zoomed video
video_path = os.path.join(stage_dir, f"stage{STAGE}_final_zoomed.mp4")
mediapy.write_video(video_path, frames, fps=50)
print(f"Episode reward: {episode_reward:.2f} | {len(frames)} frames")
print(f"Saved to {video_path}")
mediapy.show_video(frames, fps=50)
```
